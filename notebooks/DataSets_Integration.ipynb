```python

gdp_total = pd.read_csv('gdp-worldbank.csv')
water_sources = pd.read_csv('improved-water-sources-vs-gdp-per-capita.csv')
population = pd.read_csv('urban-and-rural-population.csv')

# Read Excel
hdi_data = pd.read_csv('human-development-index.csv')

governance = pd.read_excel('wgidataset_with_sourcedata.xlsx')
disasters_natural = pd.read_excel('public_emdat_custom_request_2025-11-21_d9b1c597-7944-4699-ab43-6d73937052e2.xlsx')
disasters_tech = pd.read_excel('public_emdat_custom_request_2025-11-21_2d8c3622-d8e3-4a77-bd4b-b72013093c9c.xlsx')
```

```python
def show_columns(df, name):
    print("="*70)
    print(f" Columns in {name}:")
    print("="*70)
    for col in df.columns:
        print(col)
    print("\n\n")



show_columns(gdp_total, "GDP Total Dataset")
show_columns(water_sources, "Water Sources Dataset")
show_columns(population, "Population Dataset")
show_columns(governance, "Governance Dataset")
show_columns(disasters_natural, "Natural Disaster Dataset")
show_columns(disasters_tech, "Technological Disaster Dataset")
```

```python
merged = gdp_total.merge(
    water_sources,
    on=["Entity", "Code", "Year"],
    how="outer"
)
```

```python
merged.head()
```

```python
merged = merged.merge(
    population[['Entity','Code','Year','Urban population','Rural population']],
    on=['Entity','Code','Year'],
    how='left'
)
```

```python
merged.head()
```

```python
merged.columns
```

```python
merged['Year'] = merged['Year'].astype(int)
hdi_data['Year'] = hdi_data['Year'].astype(int)
merged['Entity'] = merged['Entity'].str.strip()
merged['Code'] = merged['Code'].str.strip()
hdi_data['Entity'] = hdi_data['Entity'].str.strip()
hdi_data['Code'] = hdi_data['Code'].str.strip()
print(set(merged['Code']) - set(hdi_data['Code']))
print(set(hdi_data['Code']) - set(merged['Code']))
```

```python
# Only select the relevant columns from HDI to avoid duplicate 'World regions' columns
hdi_subset= hdi_data[['Entity', 'Code', 'Year', 'Human Development Index']]

# Merge with main dataset
merged = merged.merge(
    hdi_subset,
    on=['Entity', 'Code', 'Year'],
    how='left'
)
```

```python
merged.columns
```

```python
merged.head()
```

```python
gov_subset = governance[['code', 'countryname', 'year', 'indicator', 'estimate']]

gov_subset = gov_subset.rename(columns={
    'code': 'Code',
    'countryname': 'Entity',
    'year': 'Year'
})
```

```python
merged = merged.merge(
    gov_subset,
    on=['Entity', 'Code', 'Year'],
    how='left'
)
```

```python
merged.head()
```

```python
merged.columns
```

```python
disasters_subset = disasters_natural[[
    'Country', 'ISO', 'Start Year', 'Region',
    'Disaster Group', 'Disaster Type',
    'Total Deaths', 'No. Injured', 'No. Affected', 'No. Homeless', 'Total Affected',
    "Total Damage ('000 US$)"
]]


disasters_subset = disasters_subset.rename(columns={
    'ISO': 'Code',
    'Country': 'Entity',
    'Start Year': 'Year'
})


```

```python

merged = merged.merge(
    disasters_subset,
    on=['Entity', 'Code', 'Year'],
    how='left'
)
```

```python
merged.columns
```

```python
merged.head()
```

```python
# Keep only the required columns from technological disasters dataset
disasters_tech_subset = disasters_tech[[
    'ISO', 'Country', 'Region', 'Start Year',
    'Disaster Type',
    'Total Deaths', 'Total Affected', 'No. Affected',
    "Total Damage ('000 US$)", 'Disaster Group'
]]

# Rename columns to match your merged dataset
disasters_tech_subset = disasters_tech_subset.rename(columns={
    'ISO': 'Code',
    'Country': 'Entity',
    'Start Year': 'Year'
})

# Merge with your main dataset
merged = merged.merge(
    disasters_tech_subset,
    on=['Entity', 'Code', 'Year'],
    how='left',  # Keep all rows from your main dataset
    suffixes=('', '_tech')  # Avoid overwriting existing disaster columns
)
```

```python
merged.columns
```

```python
# 1. Fill missing Region using Region_tech
if 'Region' in merged.columns and 'Region_tech' in merged.columns:
    merged['Region'] = merged['Region'].fillna(merged['Region_tech'])

merged = merged.drop(
    columns=[
        'Region_tech',
        'World regions according to OWID',
    ],
    errors='ignore'
)
```
